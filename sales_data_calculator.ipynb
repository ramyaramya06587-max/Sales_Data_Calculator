```python
sales=[1200, 2500, 1800, 3200, 1500, 2800, 2100, 3500, 1600, 2300]

print("-" * 40)
print("           SALES DATA CALCULATOR            ")
print("-" * 40)

print("\nSales Data:", sales)
print("Number of Sales:", len(sales))

total_sales = sum(sales)
print("\nTotal Sales:", total_sales)

average_sales = total_sales/len(sales)
print("\nAverage Sales:", average_sales)

highest_sale = max(sales)
lowest_sale = min(sales)

print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)

print("\n" + "-" * 40)
print("            SALES ANALYSIS REPORT            ")
print("-" * 40)

print("Total Number of Sales:", len(sales))
print("Total Sales:" ,total_sales)
print("Average Sales:" ,average_sales)
print("Highest Sale:" ,highest_sale)
print("Lowest Sale:" ,lowest_sale)

print("\n" + "-" * 40)
print("             RESULTS AND CONCLUSION            ")
print("-" * 40)

print("The sales data was analyzed successfully using Python.")
print("Total sales represent the sum of all sales values.")
print("Average sales represent the mean value of sales.")
print("The highest and lowest sales were identified successfully.")

print("\nAnalysis completed successfully!")
print("-" * 40)
```
